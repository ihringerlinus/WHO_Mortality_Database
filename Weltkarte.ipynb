```python
import plotly.express as px
import pandas as pd

data = {
    'Country': ['Germany', 'France', 'Brazil', 'Japan', 'United States'],
    'Value': [135943, 95832, 307521, 139281, 160718]
}
df = pd.DataFrame(data)

# Erstelle eine Choroplethenkarte
fig = px.choropleth(
    df,
    locations='Country',                # Name der Länderspalte
    locationmode='country names',       # Modus: benutze Ländernamen
    color='Value',                      # Farbe nach diesem Wert
    color_continuous_scale='Viridis',   # Farbskala (alternativ: 'Plasma', 'Reds', 'Blues' etc.)
    title='Weltkarte nach Wert pro Land'
)

fig.write_html("weltkarte.html")
```
